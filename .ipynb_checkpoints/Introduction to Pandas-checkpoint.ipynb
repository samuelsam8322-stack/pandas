```python
import pandas as pd
import numpy as np

data = {
    "ID": [1, 2, 3, 4, 4],
    "Name": ["Sam", "John", "Anu", "Ravi", "Ravi"],
    "Age": [22, 25, np.nan, 28, 28],
    "City": ["Chennai", "Delhi", "Mumbai", None, None]
}

df = pd.DataFrame(data)

print(df.shape)
print(df.info())
print(df.describe())
print(df.head())
print(df.tail())

df["Salary"] = [30000, 40000, 35000, 45000, 45000]

print(df.isnull())
print(df.isnull().sum())

print(df.dropna())

print(df.duplicated())

print(df.drop_duplicates())

print(df.drop_duplicates(keep='last'))

print(df.drop_duplicates(subset=["Name"]))
```

Types of Merge

```python
pd.merge(df1, df2, on="id", how="inner")                                   # inner merge
```

```python
pd.merge(df1, df2, on="id", how="left")                                  # left merge
```

```python
pd.merge(df1, df2, on="id", how="right")                                 # right merge
```

```python
pd.merge(df1, df2, on="id", how="outer")                                # outer merge
```

```python
import pandas as pd
import numpy as np

data = {
    "id": [1, 2, 3, 4],
    "name": ["sam", "john", "anu", np.nan],
    "age": [22, 25, np.nan, 28],
    "city": ["chennai", "delhi", "mumbai", "pune"]
}

df = pd.DataFrame(data)

# fillna
df["age"] = df["age"].fillna(0)
df["name"] = df["name"].fillna("unknown")

# astype
df["age"] = df["age"].astype(int)

# columns
print(df.columns)

# upper
df.columns = df.columns.str.upper()

# lower
df.columns = df.columns.str.lower()

# rename
df.rename(columns={"name": "student_name"}, inplace=True)

print(df)

# merge
df1 = pd.DataFrame({
    "id": [1, 2, 3],
    "name": ["sam", "john", "anu"]
})

df2 = pd.DataFrame({
    "id": [1, 2, 3],
    "salary": [30000, 40000, 50000]
})

result = pd.merge(df1, df2, on="id")

print(result)
```

# OUTLIERS, SKEWNESS, KURTOSIS

# 1. OUTLIERS:

```python
An outlier is a value that is very different from other values in the dataset.

# Two types: IQR -InterQuartile Range
#            Z-score

    Detect Outliers Using IQR Method
#Formula:
IQR = Q3 - Q1
Lower Limit = Q1 - 1.5 × IQR
Upper Limit = Q3 + 1.5 × IQR
```

```python
Q1 = data['Employee_age'].quantile(0.25)
Q3 = data['Employee_age'].quantile(0.75)
```

```python
Q1,Q3
```

```python
IQR = Q3 - Q1
IQR
```

```python
lower_outlier = Q1 - 1.5*IQR
upper_outlier = Q3 + 1.5*IQR

lower_outlier
upper_outlier
```

```python
data[(data['Employee_age']< lower_outlier) | (data['Employee_age']>upper_outlier)]
```

```python
data = data[(data['Employee_age']> lower_outlier) & (data['Employee_age']<upper_outlier)]
data
```

# 2. SKEWNESS

```python
Skewness tells whether data is symmetric or not.( or To know about data spread)

# Types of Skewness
Positive Skew (Right Skew) - Tail is towards the right side.
Negative Skew (Left Skew) - Tail is towards the left side.
Symmetric Distribution - Mean ≈ Median
                                                         df.skew()
```

```python
# print(df["Marks"].skew())
data["Employee_age"].skew()
```

```python
data['Current_Salary'].skew()
```

 # 3. KURTOSIS

```python
Kurtosis measures the peakedness of data and presence of outliers (it measures the 'Tailedness'.It tells you how much of
your data lies in the tail).

#Types of Kurtosis:
1.Leptokurtic
      High peak
      More outliers
      Kurtosis > 0
2.Mesokurtic
      Normal distribution
      Kurtosis = 0
3.Platykurtic
     Flat distribution
     Fewer outliers
     Kurtosis < 0
                                                      print(df["Marks"].kurt())
```

```python
data["Employee_age"].kurt()
```

```python
data['Current_Salary'].kurt()
```

# GROUP BY

```python
groupby() is used to group similar data together and perform operations like:

sum()
mean()
count()
max()
min()

It works similar to SQL GROUP BY.
```

```python
df.groupby("column_name")
```

```python
data.groupby('Department')['Current_Salary'].mean()
```

Multiple Group By

```python
data.groupby('Department').agg({'Current_Salary':['mean','max','min'], 'years_experience':'sum'})
```

```python
import matplotlib.pyplot as plt
data.groupby('Department').agg({'Current_Salary':['mean','max','min'], 'years_experience':'sum'}).plot(kind = 'hist', title = 'Yrs Exp')
plt.ylabel('years_experience')
plt.show()                                                                                                       
```
